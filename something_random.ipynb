```python
from pyspark.sql.functions import lit

class data_quality:
  def __init__(self, source_system, task):
    self.source_system = source_system
    self.task = task

  def add_audit_columns(self, df):
    df = df.withColumn("source_system", lit(self.source_system))
    df = df.withColumn("task", lit(self.task))
    print('Audit columns added. New dataframe created')
    return df
  
  def add_filter_dataframe(self, df):
    new_df = self.add_audit_columns(df)
    new_df = new_df.filter(df.id == '1')
    return new_df  
```
